# Exploratory Data Analysis (EDA) Project

The easiest approach is to reuse the same Student Performance dataset from Project 2 and perform a complete EDA on it.

# What is EDA?

EDA means understanding the data before building a model.

You answer questions like:

How many students passed?
How are marks distributed?
Does attendance affect results?
Do study hours influence marks?
Which feature is most important?

EDA = Statistics + Visualization + Insights

```python
#Load Dataset
import pandas as pd
dataset = pd.read_csv("C:/Users/Uma Maheswara Rao/Desktop/EDA_Project/dataset/student_data.csv")
print(dataset.head())
print(dataset.shape)
```

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Dataset Information
print(dataset.shape)
print(dataset.columns)
print(dataset.info())
```

# Target Variable Distribution

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='Result',
    data=dataset
)

plt.title("Pass vs Fail Distribution")

plt.show()
```

# Study Hours Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(
    dataset['Study_Hours'],
    bins=10,
    kde=True
)

plt.title("Study Hours Distribution")

plt.show()
```

# Attendance Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(
    dataset['Attendance'],
    bins=10,
    kde=True
)

plt.title("Attendance Distribution")

plt.show()
```

# Previous Marks Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(
    dataset['Previous_Marks'],
    bins=10,
    kde=True
)

plt.title("Previous Marks Distribution")

plt.show()
```

# Box Plot

Detect outliers.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=dataset['Previous_Marks']
)

plt.title("Outlier Detection")

plt.show()
```

# Study Hours vs Previous Marks

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Study_Hours',
    y='Previous_Marks',
    hue='Result',
    data=dataset
)

plt.title("Study Hours vs Previous Marks")

plt.show()
```

# Attendance vs Marks

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Attendance',
    y='Previous_Marks',
    hue='Result',
    data=dataset
)

plt.title("Attendance vs Marks")

plt.show()
```

# Correlation Analysis

Most important EDA visualization.

Insight

Look for values near:

1   = Strong Positive

-1  = Strong Negative

0   = No Relationship

```python
numeric_columns = dataset.select_dtypes(
    include='number'
)

plt.figure(figsize=(8,6))

sns.heatmap(
    numeric_columns.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

# Pair Plot

Shows all relationships together.

```python
sns.pairplot(
    dataset,
    hue='Result'
)

plt.show()
```

# Average Marks by Result

```python
dataset.groupby(
    'Result'
)['Previous_Marks'].mean()

#Visualization:
sns.barplot(
    x='Result',
    y='Previous_Marks',
    data=dataset
)

plt.show()
```
